```python
# creating bands  of price
bins = [0, 50, 100, 200, 500, float("inf")]
labels = [
    "$0-50",
    "$51-100",
    "$101-200",
    "$201-500",
    "Above $500"]
df["price_band"] = pd.cut(
    df["price"],
    bins=bins,
    labels=labels
)
```

```python
# dividing availibility inot 4 equal column
df["avl_4grp"] = pd.cut(
    df["availability_365"],
    bins=4)
```

```python
# pivoted table of each band showing median price of each band
median_table = pd.pivot_table(
    df,
    index="price_band",
    columns="avl_4grp",
    values="price",
    aggfunc="median",
    observed=False)
print(median_table)
```

```python
#pivoted table for showing listing cont median
listing_count = pd.pivot_table(
    df,
    index="price_band",
    columns="avl_4grp",
    values="id",
    aggfunc="count",
    observed=False)
print(listing_count)
```
