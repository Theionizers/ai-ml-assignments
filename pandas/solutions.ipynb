```python
import pandas as pd
import numpy as np
df=pd.read_csv("AB_NYC_2019.csv",parse_dates=['last_review'])
q1 = df.groupby("room_type")["price"].transform("quantile", 0.25)
q3 = df.groupby("room_type")["price"].transform("quantile", 0.75)
iqr = q3 - q1
keep = (df["price"] >= q1 - 1.5 * iqr) & (df["price"] <= q3 + 1.5 * iqr)
df[keep]
```

```python
g=df.groupby("host_id")
g.agg(
    { 'price':["mean","median"],
      "availability_365":"sum"
        }
)
```

```python
ans1=g.size()
ans1=[ans1[ans1>=5]]
ans1
```

```python
ans2=g.size().sort_values(ascending=False)
ans2.head(10)
```
